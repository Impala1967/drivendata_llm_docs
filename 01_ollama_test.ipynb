```python
import ollama
```

```python
response = ollama.chat(
    model="gemma3:4b",
    messages=[
        {"role":"user",
         "content":"Explain what a research paper is in exactly 3 sentences"
        }
    ]
)
print(response["message"]["content"])
```

```python
def ask_llm(prompt):
    response=ollama.chat(
        model="gemma3:4b",
        messages=[
            {"role":"user",
             "content":prompt
            }
        ]
    )
    return response["message"]["content"]
```

```python
answer=ask_llm("Explain the difference between a research paper and a research paper abstract.")
print(answer)
```
